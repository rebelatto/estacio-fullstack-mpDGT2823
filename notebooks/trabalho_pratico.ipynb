**Contexto:** como Analista de Dados, recebi um conjunto de dados que precisa ser tratado para depois ser usado em análise e mineração. O conjunto (`data/dados.csv`) tem as colunas `ID;Duration;Date;Pulse;Maxpulse;Calories` e contém dados inválidos:

- valores nulos em `Calories` (linhas 18 e 28);
- valor nulo em `Date` (linha 22);
- data em formato errado em `Date` (linha 26: `20201226`).

## 1. Leitura do CSV

O arquivo usa `;` como separador, por isso o parâmetro `sep` é necessário. Também informo a engine `python` e o encoding `utf-8`.

```python
import pandas as pd
import numpy as np

dados = pd.read_csv('../data/dados.csv', sep=';', engine='python', encoding='utf-8')
```

## 2. Verificação da importação

### 2.1 Informações gerais

```python
dados.info()
```

### 2.2 Primeiras e últimas N linhas

```python
print(dados.head(10))
```

```python
print(dados.tail(10))
```

## 3. Cópia do conjunto de dados

Todo o tratamento é feito numa cópia, para preservar os dados originais.

```python
dados_tratados = dados.copy()
```

## 4. Substituir os nulos de `Calories` por 0

```python
dados_tratados['Calories'] = dados_tratados['Calories'].fillna(0)

print(dados_tratados.to_string())
```

As linhas 18 e 28 agora têm `Calories = 0.0`.

## 5. Substituir os nulos de `Date` por `'1900/01/01'`

```python
dados_tratados['Date'] = dados_tratados['Date'].fillna('1900/01/01')

print(dados_tratados.to_string())
```

## 6. Converter `Date` para datetime (1ª tentativa)

Como o roteiro previa, a conversão falha: o pandas deduz o formato `'%Y/%m/%d'` (com as aspas simples que existem no arquivo) e o valor `1900/01/01` não segue esse padrão. Uso `try/except` para exibir o erro sem interromper o notebook.

```python
try:
    dados_tratados['Date'] = pd.to_datetime(dados_tratados['Date'])
except ValueError as erro:
    print('Erro encontrado:')
    print(erro)
```

## 7. Substituir `'1900/01/01'` por `NaN` e tentar de novo

```python
dados_tratados['Date'] = dados_tratados['Date'].replace('1900/01/01', np.nan)

try:
    dados_tratados['Date'] = pd.to_datetime(dados_tratados['Date'])
except ValueError as erro:
    print('Erro encontrado:')
    print(erro)

print()
print(dados_tratados.to_string())
```

Novo erro, agora com o valor `20201226` (linha 26), que não corresponde ao formato `'%Y/%m/%d'`.

## 8. Corrigir o valor `20201226` com `replace` + `to_datetime`

Troco especificamente esse texto pela data convertida com `to_datetime`, informando o formato dele (`%Y%m%d`).

```python
dados_tratados['Date'] = dados_tratados['Date'].replace(
    '20201226', pd.to_datetime('20201226', format='%Y%m%d')
)

print(dados_tratados.loc[26])
```

## 9. Converter toda a coluna `Date` para datetime

Agora todos os valores de texto seguem o formato `'%Y/%m/%d'` (com aspas), então informo esse formato explicitamente.

```python
dados_tratados['Date'] = pd.to_datetime(dados_tratados['Date'], format="'%Y/%m/%d'")

print(dados_tratados.to_string())
print()
print('Tipo da coluna Date:', dados_tratados['Date'].dtype)
```

A linha 22 aparece como `NaT` (data nula), que será removida no próximo passo.

## 10. Remover registros com valores nulos

Apenas a coluna `Date` ainda tem um nulo (linha 22), então ela é usada como base para a remoção.

```python
dados_tratados = dados_tratados.dropna(subset=['Date'])

print(dados_tratados.to_string())
```

## 11. Conferência final

```python
dados_tratados.info()

print()
print('Linhas antes do tratamento:', len(dados))
print('Linhas depois do tratamento:', len(dados_tratados))
print('Valores nulos restantes:', dados_tratados.isnull().sum().sum())
```

## Conclusão

O conjunto foi tratado com pandas:

- nulos de `Calories` substituídos por 0;
- coluna `Date` convertida para o tipo `datetime64`, incluindo a correção do valor fora do padrão (`20201226`);
- o registro sem data (linha 22) foi removido.

O resultado final tem 31 linhas, nenhum valor nulo e tipos adequados em todas as colunas, ficando pronto para análise.
