## Microatividade 1 — Ler um arquivo CSV com pandas

Importa a biblioteca, cria uma variável e atribui a ela o conteúdo do CSV, informando o separador (`;`), a engine (`python`) e o encoding.

```python
import pandas as pd

df = pd.read_csv('../data/OnlineRetail.csv', sep=';', engine='python', encoding='utf-8')

print(df)
```

## Microatividade 2 — Criar um subconjunto de dados

Nova variável contendo apenas 4 colunas do conjunto original.

```python
subconjunto = df[['InvoiceNo', 'CustomerID', 'Quantity', 'UnitPrice']]

print(subconjunto)
```

## Microatividade 3 — Configurar o número máximo de linhas exibidas

Define a opção `display.max_rows` para 9999 e imprime o conjunto original com `to_string()`.

> Observação: o `max_rows` controla quantas linhas o pandas mostra na exibição normal de um DataFrame (por exemplo, `print(df)` ou `display(df)`). Já o `to_string()` converte o DataFrame inteiro em texto e ignora essa opção, por isso a saída abaixo mostra todas as 14.999 linhas.

```python
pd.set_option('display.max_rows', 9999)

print(df.to_string())
```

### Por que usar `head()` junto com `to_string()`

A opção `display.max_rows` define quantas linhas o pandas mostra na exibição automática de um DataFrame, como no `print(df)`. O método `to_string()`, porém, ignora essa opção e converte a tabela inteira em texto. Por isso, só configurar o limite não basta: o `to_string()` exibiria todas as 14.999 linhas.

Para respeitar o limite configurado, o valor é lido de volta com `pd.get_option('display.max_rows')` e usado no `head()`. O `head()` corta a tabela antes da conversão, e o `to_string()` passa a exibir apenas as 9.999 primeiras linhas, em sequência. Assim, se o valor do `set_option` mudar, a quantidade de linhas exibidas acompanha a configuração.

```python
pd.set_option('display.max_rows', 9999)

print(df.head(pd.get_option('display.max_rows')).to_string())
```

## Microatividade 5 — Informações gerais sobre o conjunto de dados

`info()` mostra colunas, total de linhas, valores não nulos, tipo de cada coluna e memória utilizada.

```python
df.info()
```

```python
print('Total de linhas:', df.shape[0])
print('Total de colunas:', df.shape[1])
print('\nDados nulos por coluna:')
print(df.isnull().sum())
print('\nTipo de dado de cada coluna:')
print(df.dtypes)
print('\nMemória utilizada (bytes):', df.memory_usage(deep=True).sum())
```

### Respostas (a partir do `info()`)

- **Total de linhas:** 14.999 (índices 0 a 14998)
- **Total de colunas:** 8 (`InvoiceNo`, `StockCode`, `Description`, `Quantity`, `InvoiceDate`, `UnitPrice`, `CustomerID`, `Country`)
- **Dados nulos:** 4.331 no total — 60 na coluna `Description` e 4.271 na coluna `CustomerID`; as demais colunas não têm nulos
- **Tipos:**
  - `Quantity` → inteiro (`int64`)
  - `CustomerID` → decimal (`float64`; o pandas usa decimal porque a coluna tem valores nulos)
  - `InvoiceNo`, `StockCode`, `Description`, `InvoiceDate`, `UnitPrice` e `Country` → texto (`str`). `UnitPrice` fica como texto porque o arquivo usa vírgula como separador decimal (ex.: `2,55`)
- **Memória:** 937,6 KB segundo o `info()`. Esse valor não conta o conteúdo completo dos textos; com `memory_usage(deep=True)`, que mede tudo, o total é 5.719.150 bytes (aprox. 5,5 MB)
